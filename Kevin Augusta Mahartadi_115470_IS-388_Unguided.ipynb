```python
!pip install xgboost
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.impute import SimpleImputer
import joblib
```

```python
BASE_DIR = os.getcwd()

covid_path = os.path.join(BASE_DIR, "data", "/content/covid_19_indonesia_time_series_all.csv")
mob_path   = os.path.join(BASE_DIR, "data", "/content/2020_ID_Region_Mobility_Report.csv")

covid = pd.read_csv(covid_path)
mob   = pd.read_csv(mob_path)
```

```python
covid = covid[['Date', 'New Cases']]
covid['Date'] = pd.to_datetime(covid['Date'])

mob = mob[
    [
        'date',
        'retail_and_recreation_percent_change_from_baseline',
        'grocery_and_pharmacy_percent_change_from_baseline',
        'transit_stations_percent_change_from_baseline',
        'workplaces_percent_change_from_baseline',
        'residential_percent_change_from_baseline',
    ]
]
mob['date'] = pd.to_datetime(mob['date'])
```

```python
merged = pd.merge(covid, mob, left_on='Date', right_on='date', how='inner')

merged = merged.drop(columns=['date', 'Date'])

print("Ukuran merge:", merged.shape)
print(merged.head())
```

```python
X = merged.drop(columns=['New Cases'])
y = merged['New Cases']

X = X.select_dtypes(include=['int64', 'float64'])

imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(X)
```

```python
from xgboost import XGBRegressor
from sklearn.model_selection import GridSearchCV, train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_imputed, y, test_size=0.2, random_state=42)

xgb_model = XGBRegressor(random_state=42)

param_grid_xgb = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [3, 5, 8]
}

grid_search_xgb = GridSearchCV(
    estimator=xgb_model,
    param_grid=param_grid_xgb,
    cv=5,
    scoring='r2',
    n_jobs=-1,
    verbose=1
)

print("Starting GridSearchCV for XGBoost Regressor")
grid_search_xgb.fit(X_train, y_train)
print("GridSearchCV completed")
```

```python
best_xgb_model = grid_search_xgb.best_estimator_

print("\nBest Hyperparameters found:")
print(grid_search_xgb.best_params_)
```

```python
y_pred_xgb = best_xgb_model.predict(X_test)

r2_xgb = r2_score(y_test, y_pred_xgb)
mse_xgb = mean_squared_error(y_test, y_pred_xgb)
mae_xgb = mean_absolute_error(y_test, y_pred_xgb)

print("\n--- Best XGBoost Model Performance on Test Set ---")
print(f"R-squared: {r2_xgb:.4f}")
print(f"MSE: {mse_xgb:.4f}")
print(f"MAE: {mae_xgb:.4f}")
```

```python
model_dir = os.path.join(os.getcwd(), "model")
os.makedirs(model_dir, exist_ok=True)
model_path = os.path.join(model_dir, "xgb_model.pkl")
joblib.dump(best_xgb_model, model_path)

print(f"\nModel Saved: {model_path}")
```
